# **Question #1** - Aggregates on User Input
---
<br/>

Write a program that takes in from the user a list of marks until the user enters a negative number. Once this is done, the program should calculate and display



1.   The number of marks
2.   The highest mark
3.   The lowest mark
4.   The average of the marks

(note: You cannot use the `max()` or `min()` functions)

<br/>


**<u>Example Run:</u>**<br/>
Mark #1? **50**<br/>
Mark #2? **60**<br/>
Mark #3? **-1**<br/>
Number of marks = 2<br/>
Highest Mark = 60<br/>
Lowest Mark = 50<br/>
Average Mark = 55<br/>
<br/>

<Br/>

><table>
### BONUS
---
Try and add some validation into the program where if the user enters a mark over 100, a message is sent to the user that lets them know that the mark is invalid (and it doesn't add the mark to the statistics - the program just loops again to get another mark).
</table>

### Your Answer:

````python
# Student Name: Todd C
# Student Number: 1037326

# Add answer below this line
entries = []
count = 0

while entries != 0:
  entry = int(input("Enter mark: "))
  if entry > 0:
    entries.append(entry)
    count += 1
    if entry >100:
      print("Invalid Grade")

  elif entry < 0:
      average = sum(entries)/len(entries)

      lowest = entries[0]  # Start with the first element
      for num in entries[1:]:
          if num < lowest:
              lowest = num

      highest = entries[0]  # Start with the first element
      for num1 in entries[1:]:
          if num1 > highest:
              highest = num1


      print("Num of Marks:",count)
      print("Highest Mark:",highest)
      print("Lowest Mark:",lowest)
      print("Average Mark:",average)
      break

````

### Your Answer:

````python
# Student Name: Todd C
# Student Number: 1037326

# Add answer below this line
balance = float(input("Balance?"))
years = float(input("years"))
rate = float(input("Interest rate?"))/100
count = 0

print("""Year      Balance     Interest   Total Int.
-------------------------------------------""")
while count < years:
  newadd = balance*rate

  print(count, balance+newadd,"      ", newadd,"                   ", (balance*newadd)/100 )
  count +=1
  balance = balance+newadd
print("test")


#type error in collab, functiong fine in other IDE
````
