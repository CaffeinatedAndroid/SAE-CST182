# **Question #1** - Sort three numbers
---
<br/>

Write a program that asks for three numbers from the user, and prints out the three numbers in ascending order.  <br/><br/>


**<u>Example Run:</u>**<br/>
Number #1? **9**<br/>
Number #1? **3**<br/>
Number #1? **5**<br/>
Result = 3 5 9<br/>
<br/>
>**Note:** *You cannot use the `max()` or `min()` functions*


### Your Answer:

```python
# Student Name:   TODD_C
# Student Number: 1037326

# Add answer below this line


num1 = int(input("Number 1? "))
num2 = int(input("Number 2? "))
num3 = int(input("Number 3? "))

if num1 > num2:
    num1, num2 = num2, num1
if num1 > num3:
    num1, num3 = num3, num1
if num2 > num3:
    num2, num3 = num3, num2

print("Result =", num1, num2, num3)
```

```python
# Student Name:   TODD_C
# Student Number: 1037326
# BMI Ranges Reference URL: https://www.health.gov.au/topics/overweight-and-obesity/bmi-and-waist?language=en

# Add answer below this line

print("______BMI CALCULATOR______\n\nWelcome, please enter your information:\n")


name = str(input("What is your name?\n"))
height = float(input("Height(M)?\n"))
weight = float(input("Weight(Kg)?\n"))
bmi = float(weight / height**2)

print("\n\nHi",name,"\nYour BMI is as followed:")
print("Weight:",weight,"Kg")
print("Height:",height,"M")
print("BMI:",bmi)

if bmi < 18.5:
  print("The given BMI is classified as Underweight")
elif bmi > 18.5:
  print("The given BMI is classified as Healthy")
elif bmi > 25:
  print("The given BMI is classified as Overweight but not obese")
elif bmi > 30:
  print("The given BMI is classified as Obese class I")
elif bmi > 35:
  print("The given BMI is classified as Obese class II")
elif bmi > 40:
  print("The given BMI is classified as Obese class III")







```

# **Question #3** - Movie tickets
---

Create a program that calculates the cost of a ticket at the movies. The program asks the person's age. If they are under 18, they get a child's ticket at \$10.00, and adults pay \$20.00. However, if a person is over 65, they are asked if they have a concession card, and if so, then receive a \$15.00 ticket. <br/><br/>

**<u>Example Run #1:</u>**<br/>
Age? **15**<br/>
Ticket = $10.00<br/><br/>

**<u>Example Run #2:</u>**<br/>
Age? **30**<br/>
Ticket = $20.00<br/><br/>

**<u>Example Run #3:</u>**<br/>
Age? **70**<br/>
Concession Card? **Yes**<br/>
Ticket = $15.00<br/><br/>

**<u>Example Run #4:</u>**<br/>
Age? **70**<br/>
Concession Card? **No**<br/>
Ticket = $20.00<br/><br/>

### Your Answer:

```python
# Student Name:   TODD_C
# Student Number: 1037326

# Add answer below this line
print("_____WELCOME TO TERMINAL CINEMAS_____")

priceAdult = float(20.00)
priceMinor = float(10.00)
priceConcession = float(15.00)
endPrefix = str("Have a great day")

age = int(input(" Please enter your age: "))


if age >= 65:
  i = str(input("Do you have a concession? (yes/no) "))
  if i == str("yes"):
    print(" Ticket $",priceConcession, "\nConcession: YES\n",endPrefix)
  else:
    print(" Ticket $",priceAdult, "\nConcession: NO\n",endPrefix)

elif age < 18:
  print(" Ticket $",priceMinor, "\n",endPrefix)

elif age > 18:
  print(" Ticket $",priceAdult, "\n",endPrefix)




```
